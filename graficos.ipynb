```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv('ecommerce_preparados (2).csv')
print(df.head().to_string())
```

```python
# Histograma
plt.figure(figsize=(10,6))
plt.hist(df['Desconto'], bins=20, color='blue', edgecolor='black')
plt.title('Histograma de Desconto')
plt.xlabel('Desconto')
plt.ylabel('Frequência')
plt.grid(True)
plt.show()
```

```python
# Gráfico de dispersão
sns.jointplot(x='Desconto', y='Preço', data=df, kind='scatter', color='blue')
plt.title('Gráfico de Dispersão - Desconto vs Preço')
plt.show()
plt.figure(figsize=(10,6))
plt.subplot(2,2,1)
plt.scatter(df['Desconto'], df['Preço'], color='blue', alpha=0.5)
plt.title('Gráfico de Dispersão - Desconto e Preço')
plt.xlabel('Desconto')
plt.ylabel('Preço')

plt.subplot(1,2,2)
plt.scatter(df['Preço'], df['Nota'], color='green', alpha=0.5)
plt.title('Gráfico de Dispersão - Preço e Nota')
plt.xlabel('Preço')
plt.ylabel('Nota')

plt.figure(figsize=(10,6))
plt.subplot(2,2,1)
plt.scatter(df['Desconto'], df['Desconto'], color='blue', alpha=0.5)
plt.title('Gráfico de Dispersão - Desconto e Desconto')
plt.xlabel('Desconto')
plt.ylabel('Desconto')

plt.hexbin(df['Desconto'], df['Nota'], gridsize=20, cmap='viridis', mincnt=1)
plt.colorbar(label='Frequência')
plt.title('Gráfico de Dispersão - Desconto e Nota')
plt.xlabel('Desconto')
plt.ylabel('Nota')
plt.show()

```

```python
# Mapa de Calor
corr = df[['Desconto', 'Preço', 'Nota']].corr()
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Mapa de Calor de Correlação Desconto, Preço e Nota')

plt.tight_layout()
plt.show()
```

```python
# Gráfico Barra
plt.figure(figsize=(8,6))
df['Gênero'].value_counts().plot(kind='bar', color=['blue', 'orange'])
plt.title('Quantidade de Clientes por Gênero')
plt.xlabel('Gênero')
plt.ylabel('Quantidade de Clientes')
plt.show()

x = df['Gênero'].value_counts().index
y = df['Gênero'].value_counts().values

plt.figure(figsize=(8,6))
plt.bar(x,y, color=['blue', 'orange'])
plt.title('Quantidade de Clientes por Gênero')
plt.xlabel('Gênero')
plt.ylabel('Quantidade de Clientes')
plt.show()
```

```python
# Gráfico de pizza
plt.figure(figsize=(10,6))
plt.pie(y, labels=x, autopct='%.1f%%', startangle=180)
plt.title('Porcentagem de Clientes por Gênero')
plt.show()
```

```python
#Grafico de Densidade
plt.figure(figsize=(10,6))
sns.kdeplot(df['Desconto'], color='blue', fill=True)
plt.title('Densidade de Desconto')
plt.xlabel('Desconto')
plt.ylabel('Densidade')
```

```python
#Grafico de Regressão
sns.regplot(x='Desconto', y='Preço', data=df, scatter_kws={'color': 'blue'}, line_kws={'color': 'red'})
plt.title('Regressão Linear - Desconto vs Preço')
plt.xlabel('Desconto')
plt.ylabel('Preço')
plt.show()
```
