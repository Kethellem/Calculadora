```python
# Importando as bibliotecas necessárias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Gerando o dataframe simulado
np.random.seed(42)  # Para resultados reproduzíveis

idades = np.random.randint(20, 60, 20)
salarios = idades * np.random.randint(80, 120, 20)
pontuacoes = salarios * np.random.uniform(0.5, 1.5, 20)
profissoes = ['Engenheiro', 'Professor', 'Médico', 'Advogado', 'Designer',
              'Analista', 'Gerente', 'Programador']

df = pd.DataFrame({
    'Idade': idades,
    'Salário': salarios,
    'Pontuação': pontuacoes.round(2)
})

df['Profissão'] = np.random.choice(profissoes, size=len(df))

# Exibir as primeiras linhas do dataframe
df.head()
```

Gráfico 1: Dispersão entre Salário e Pontuação (Limite de Crédito)

Este gráfico foi selecionado para verificar a relação entre o salário do cliente e seu limite de crédito (pontuação).
Observa-se que, de forma geral, quanto maior o salário do cliente, maior tende a ser seu limite de crédito. Isso é coerente, pois clientes com maior renda possuem maior capacidade financeira, o que reduz o risco para as instituições financeiras.

```python
plt.figure(figsize=(8, 6))
plt.scatter(df['Salário'], df['Pontuação'], color='purple', alpha=0.7)
plt.title('Relação entre Salário e Limite de Crédito (Pontuação)', fontsize=14)
plt.xlabel('Salário', fontsize=12)
plt.ylabel('Pontuação (Limite de Crédito)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

Gráfico 2: Boxplot da Pontuação por Profissão

Este boxplot foi escolhido para analisar como diferentes profissões influenciam o limite de crédito. Percebe-se que algumas profissões apresentam limites de crédito medianos mais altos. Profissões como Médico, Engenheiro e Gerente se destacam, indicando que ocupações com maior estabilidade financeira ou alta renda costumam receber limites mais elevados. Por outro lado, profissões com renda mais variável apresentam uma pontuação mais distribuída ou menor

```python
plt.figure(figsize=(10, 6))
df.boxplot(column='Pontuação', by='Profissão')
plt.title('Limite de Crédito (Pontuação) por Profissão')
plt.suptitle('')  # Remove o título automático do pandas
plt.xlabel('Profissão')
plt.ylabel('Pontuação')
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

ráfico 3: Média da Pontuação por Faixa de Idade

Este gráfico permite observar como a idade influencia o limite de crédito.O gráfico mostra que clientes na faixa etária de 31 a 50 anos tendem a possuir os maiores limites de crédito. Isso pode ser explicado pelo fato de que, nessa fase da vida, as pessoas geralmente estão no auge de sua vida profissional, com maior estabilidade de emprego e renda. Já os mais jovens (21-30) e os mais velhos (51-60) apresentam limites um pouco menores, seja pela menor experiência financeira ou pela proximidade da aposentadoria.

```python
# Criando faixas etárias
faixas_idade = pd.cut(df['Idade'], bins=[20, 30, 40, 50, 60], labels=['21-30', '31-40', '41-50', '51-60'])

# Calculando a média da pontuação por faixa etária
media_pontuacao_idade = df.groupby(faixas_idade)['Pontuação'].mean()

# Plotando o gráfico
plt.figure(figsize=(8, 6))
plt.bar(media_pontuacao_idade.index, media_pontuacao_idade, color='skyblue')
plt.title('Média do Limite de Crédito por Faixa de Idade')
plt.xlabel('Faixa de Idade')
plt.ylabel('Pontuação (Limite de Crédito)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
